# 🔧 PART 2: Utility Functions (Using Week 02 Concepts)

```python
print("--- Utility Functions ---")

def train_test_split(data, test_size=0.2, shuffle=False, random_state=None):
    """
    Split data into training and testing sets.
    Uses default parameters for flexibility.
    
    Returns:
        X_train, X_test, y_train, y_test
    """
    if shuffle:
        import random
        if random_state is not None:
            random.seed(random_state)
        data = list(data)
        random.shuffle(data)
    
    split_idx = int(len(data) * (1 - test_size))
    train_data = data[:split_idx]
    test_data = data[split_idx:]
    
    X_train = [sample[0] for sample in train_data]
    y_train = [sample[1] for sample in train_data]
    X_test = [sample[0] for sample in test_data]
    y_test = [sample[1] for sample in test_data]
    
    return X_train, X_test, y_train, y_test

def get_stats(*values):
    """Calculate stats using *args."""
    if len(values) == 0:
        return None, None, None
    return min(values), max(values), sum(values) / len(values)

# Test train_test_split
X_train, X_test, y_train, y_test = train_test_split(dataset, test_size=0.25)
print(f"Train: {len(X_train)} samples")
print(f"Test: {len(X_test)} samples")
```

---
# 🤖 PART 3: Model Class (Using Functions with **kwargs)

```python
print("--- KNN Model with Flexible Config ---")

def euclidean_distance(point1, point2):
    """Calculate Euclidean distance."""
    total = 0
    for i in range(len(point1)):
        total += (point1[i] - point2[i]) ** 2
    return total ** 0.5

def create_knn_model(**config):
    """
    Create a KNN model with flexible configuration.
    Uses **kwargs for flexible hyperparameters.
    
    Config options:
        k: Number of neighbors (default: 3)
        weights: 'uniform' or 'distance' (default: 'uniform')
        verbose: Print progress (default: False)
    """
    # Set defaults
    model = {
        "k": config.get("k", 3),
        "weights": config.get("weights", "uniform"),
        "verbose": config.get("verbose", False),
        "is_fitted": False,
        "training_data": None
    }
    
    print(f"Created KNN model:")
    for key, value in model.items():
        if key not in ["training_data"]:
            print(f"  {key}: {value}")
    
    return model

# Create model with custom config
model = create_knn_model(k=5, weights="uniform", verbose=True)
```

```python
def fit(model, X_train, y_train):
    """Fit the KNN model on training data."""
    model["training_data"] = list(zip(X_train, y_train))
    model["is_fitted"] = True
    
    if model["verbose"]:
        print(f"Model fitted on {len(X_train)} samples")
    
    return model

def predict_one(model, test_point):
    """Predict label for a single point."""
    if not model["is_fitted"]:
        raise ValueError("Model not fitted!")
    
    # Calculate distances
    distances = []
    for train_features, train_label in model["training_data"]:
        dist = euclidean_distance(test_point, train_features)
        distances.append((dist, train_label))
    
    # Sort and get k nearest
    distances.sort()
    k_nearest = distances[:model["k"]]
    
    # Vote
    label_counts = {}
    for dist, label in k_nearest:
        weight = 1 if model["weights"] == "uniform" else 1 / (dist + 0.001)
        label_counts[label] = label_counts.get(label, 0) + weight
    
    # Return best label
    best_label = max(label_counts, key=label_counts.get)
    return best_label

def predict(model, X_test):
    """Predict labels for multiple points."""
    return [predict_one(model, x) for x in X_test]

# Fit the model
model = fit(model, X_train, y_train)
```

# 📈 PART 4: Evaluation Functions

```python
print("--- Evaluation Functions ---")

def calculate_metrics(y_actual, y_predicted, *extra_labels, verbose=True, **options):
    """
    Calculate multiple metrics.
    Demonstrates *args and **kwargs together.
    
    Args:
        y_actual: Actual labels
        y_predicted: Predicted labels
        *extra_labels: Additional label sets for comparison
        verbose: Print results
        **options: Additional options (e.g., round_digits=2)
    
    Returns:
        Dictionary of metrics
    """
    round_digits = options.get("round_digits", 4)
    
    # Accuracy
    correct = sum(1 for a, p in zip(y_actual, y_predicted) if a == p)
    accuracy = round(correct / len(y_actual), round_digits)
    
    # Per-class accuracy
    unique_labels = set(y_actual)
    per_class = {}
    for label in unique_labels:
        label_correct = sum(1 for a, p in zip(y_actual, y_predicted) if a == label and a == p)
        label_total = sum(1 for a in y_actual if a == label)
        per_class[label] = round(label_correct / label_total if label_total > 0 else 0, round_digits)
    
    metrics = {
        "accuracy": accuracy,
        "per_class": per_class,
        "total_samples": len(y_actual),
        "correct": correct
    }
    
    if verbose:
        print(f"\nEvaluation Results:")
        print(f"  Accuracy: {accuracy:.2%}")
        print(f"  Correct: {correct}/{len(y_actual)}")
        print(f"  Per-class:")
        for label, acc in per_class.items():
            print(f"    {label}: {acc:.2%}")
    
    return metrics

# Make predictions
predictions = predict(model, X_test)

# Calculate metrics
metrics = calculate_metrics(y_test, predictions, round_digits=4)
```

# 🔬 PART 5: Hyperparameter Tuning

```python
print("--- Hyperparameter Tuning ---")

def grid_search(X_train, y_train, X_test, y_test, **param_grid):
    """
    Search for best hyperparameters.
    Uses **kwargs for flexible parameter grid.
    
    Args:
        X_train, y_train: Training data
        X_test, y_test: Testing data
        **param_grid: Parameter values to try (e.g., k=[1,3,5])
    
    Returns:
        best_params, best_accuracy, all_results
    """
    results = []
    best_accuracy = 0
    best_params = {}
    
    # Get k values to try
    k_values = param_grid.get("k", [3])
    weights_options = param_grid.get("weights", ["uniform"])
    
    print(f"Testing {len(k_values) * len(weights_options)} configurations...\n")
    
    for k in k_values:
        for weights in weights_options:
            # Create and train model
            model = create_knn_model(k=k, weights=weights, verbose=False)
            model = fit(model, X_train, y_train)
            
            # Predict and evaluate
            preds = predict(model, X_test)
            correct = sum(1 for a, p in zip(y_test, preds) if a == p)
            accuracy = correct / len(y_test)
            
            result = {"k": k, "weights": weights, "accuracy": accuracy}
            results.append(result)
            
            print(f"  k={k}, weights={weights}: {accuracy:.2%}")
            
            if accuracy > best_accuracy:
                best_accuracy = accuracy
                best_params = {"k": k, "weights": weights}
    
    return best_params, best_accuracy, results

# Run grid search
best_params, best_acc, all_results = grid_search(
    X_train, y_train, X_test, y_test,
    k=[1, 3, 5, 7],
    weights=["uniform", "distance"]
)

print(f"\n🏆 Best Parameters: {best_params}")
print(f"🏆 Best Accuracy: {best_acc:.2%}")
```

# 🏆 PART 6: Complete Pipeline Function

```python
print("=" * 50)
print("COMPLETE ML PIPELINE")
print("=" * 50)

def run_ml_experiment(data, test_size=0.2, verbose=True, **model_config):
    """
    Run a complete ML experiment.
    
    Combines all Week 01 + Week 02 concepts:
    - Default parameters
    - Multiple returns  
    - *args and **kwargs
    - Loops and conditionals
    - Functions and scope
    
    Args:
        data: Dataset as list of (features, label) tuples
        test_size: Test set proportion (default: 0.2)
        verbose: Print progress (default: True)
        **model_config: Model hyperparameters
    
    Returns:
        results dictionary with model, predictions, metrics
    """
    if verbose:
        print("\n[1] Splitting data...")
    X_train, X_test, y_train, y_test = train_test_split(data, test_size=test_size)
    if verbose:
        print(f"    Train: {len(X_train)}, Test: {len(X_test)}")
    
    if verbose:
        print("\n[2] Creating model...")
    model = create_knn_model(**model_config, verbose=False)
    if verbose:
        print(f"    Config: k={model['k']}, weights={model['weights']}")
    
    if verbose:
        print("\n[3] Training...")
    model = fit(model, X_train, y_train)
    if verbose:
        print(f"    Fitted on {len(X_train)} samples")
    
    if verbose:
        print("\n[4] Predicting...")
    predictions = predict(model, X_test)
    
    if verbose:
        print("\n[5] Evaluating...")
    metrics = calculate_metrics(y_test, predictions, verbose=verbose)
    
    return {
        "model": model,
        "predictions": predictions,
        "actuals": y_test,
        "metrics": metrics,
        "config": model_config
    }

# Run the complete experiment
results = run_ml_experiment(
    dataset,
    test_size=0.25,
    verbose=True,
    k=3,
    weights="uniform"
)

print("\n" + "=" * 50)
print(f"EXPERIMENT COMPLETE!")
print(f"Final Accuracy: {results['metrics']['accuracy']:.2%}")
print("=" * 50)
```
