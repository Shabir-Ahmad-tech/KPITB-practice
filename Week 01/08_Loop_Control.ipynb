## 📖 Loop Control Keywords

| Keyword | Action | Use Case |
|---------|--------|----------|
| `break` | Exit loop completely | Stop when condition met |
| `continue` | Skip current iteration | Filter/skip bad data |
| `pass` | Do nothing (placeholder) | Empty blocks |

---
## 📖 1. Break - Exit the Loop

````python
print("--- 1. Break Statement ---")

# Stop searching when item is found
inventory = ["potion", "shield", "key", "sword", "gem"]
target = "key"

for item in inventory:
    print(f"Checking: {item}")
    if item == target:
        print(f"  ✓ Found {target}!")
        break  # Stop the loop

print("Search complete")
````

### 🤖 AI/ML Application: Early Stopping

````python
# Early stopping when loss stops improving
max_epochs = 100
target_loss = 0.1
patience = 3  # Stop after 3 epochs without improvement

best_loss = 1.0
no_improve_count = 0

# Simulated losses
losses = [0.9, 0.7, 0.5, 0.4, 0.35, 0.35, 0.35, 0.35, 0.34]

for epoch, loss in enumerate(losses, 1):
    print(f"Epoch {epoch}: Loss = {loss}")
    
    # Check for target reached
    if loss <= target_loss:
        print(f"✓ Target loss reached!")
        break
    
    # Check for improvement
    if loss < best_loss:
        best_loss = loss
        no_improve_count = 0
    else:
        no_improve_count += 1
        if no_improve_count >= patience:
            print(f"⚠️ Early stopping! No improvement for {patience} epochs")
            break
````

## 📖 2. Continue - Skip This Round

````python
print("--- 2. Continue Statement ---")

# Skip dead enemies
enemies = [
    {"name": "Goblin", "hp": 50},
    {"name": "Skeleton", "hp": 0},  # Dead
    {"name": "Orc", "hp": 80},
    {"name": "Ghost", "hp": 0},  # Dead
    {"name": "Dragon", "hp": 200},
]

print("Attacking living enemies:")
for enemy in enemies:
    if enemy["hp"] <= 0:
        continue  # Skip dead enemies
    print(f"  Attacking {enemy['name']} (HP: {enemy['hp']})")
````

### 🤖 AI/ML Application: Filtering Invalid Data

````python
# Skip invalid samples during data processing
raw_data = [
    {"age": 25, "income": 50000},
    {"age": -5, "income": 60000},   # Invalid age
    {"age": 35, "income": 75000},
    {"age": 45, "income": None},     # Missing income
    {"age": 30, "income": 55000},
]

valid_samples = []
skipped = 0

for sample in raw_data:
    # Skip invalid age
    if sample["age"] < 0:
        print(f"Skipping: Invalid age {sample['age']}")
        skipped += 1
        continue
    
    # Skip missing values
    if sample["income"] is None:
        print(f"Skipping: Missing income")
        skipped += 1
        continue
    
    valid_samples.append(sample)

print(f"\nValid samples: {len(valid_samples)}")
print(f"Skipped: {skipped}")
````

## 📖 4. Loop Else Clause

````python
print("--- 4. Loop Else ---")

# Else runs only if loop completes without break

# Example 1: Item found (else NOT executed)
inventory = ["sword", "key", "potion"]
target = "key"

print(f"Searching for {target}:")
for item in inventory:
    if item == target:
        print(f"  Found {target}!")
        break
else:
    print(f"  {target} not found")

# Example 2: Item NOT found (else IS executed)
target = "bow"
print(f"\nSearching for {target}:")
for item in inventory:
    if item == target:
        print(f"  Found {target}!")
        break
else:
    print(f"  {target} not found")
````

---
## 📖 5. Combining Break and Continue

````python
print("--- 5. Combined Example ---")

# Process until we find 3 valid items
items = [0, 5, 0, 10, 0, 15, 20, 25, 30]  # 0s are invalid
valid_count = 0
target_count = 3
processed = []

for item in items:
    # Skip zeros
    if item == 0:
        print(f"Skipping invalid item: {item}")
        continue
    
    # Process valid item
    processed.append(item)
    valid_count += 1
    print(f"Processed: {item} ({valid_count}/{target_count})")
    
    # Stop when we have enough
    if valid_count >= target_count:
        print("Target reached!")
        break

print(f"\nResult: {processed}")
````

# 📝 Key Takeaways

| Keyword | Effect |
|---------|--------|
| `break` | Exit loop immediately |
| `continue` | Skip to next iteration |
| `pass` | Do nothing (placeholder) |
| `else` | Runs if loop completes |

**ML Connections:**
- Early stopping: `break` when loss stops improving
- Data filtering: `continue` to skip invalid samples
- Placeholder models: `pass` for TODO
