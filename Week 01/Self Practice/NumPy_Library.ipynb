# 🟢 LEVEL 1: The Foundation

**Goal:** Understand why we ditch Python lists for NumPy arrays.

## 1.1 Speed: The "Why"

ML models process millions of numbers. Python loops are too slow.

```python
size = 1_000_000
py_list = list(range(size))
np_arr = np.arange(size)

# Timer: Multiply by 2
start = time.time()
py_res = [x * 2 for x in py_list]
print(f"Python List: {time.time() - start:.4f} sec")

start = time.time()
np_res = np_arr * 2
print(f"NumPy Array: {time.time() - start:.6f} sec")
print("🚀 NumPy is ~10-50x faster!")
```

## 1.2 Creating Arrays & Data Types

**Critical for ML:** Neural networks usually use **`float32`** (not `float64`) to save GPU memory.

```python
lst = [1, 2, 3]

# Default (int or float64)
arr_default = np.array(lst)
print(f"Default: {arr_default.dtype}")

# ML Standard (float32)
arr_ml = np.array(lst, dtype=np.float32)
print(f"ML Type: {arr_ml.dtype}")

# Common constructors
zeros = np.zeros((3, 3))       # Init weights to 0
ones = np.ones((2, 2))         # Init weights to 1
ranges = np.arange(0, 10, 2)   # [0, 2, 4...]
linear = np.linspace(0, 1, 5)  # 5 steps from 0 to 1

print(f"Linear Space: {linear}")
```

## 1.3 Element-wise Math (Vectorization)

No loops! Operations apply to *every* element.

```python
prices = np.array([100, 200, 300])
tax = 0.1

total = prices * (1 + tax)
print(f"With Tax: {total}")
```

## 1.4 Aggregations (Stats)

Standard ML metrics often just need simple stats.

```python
errors = np.array([0.5, 0.2, 0.1, 0.8])

print(f"Max Error: {errors.max()}")
print(f"Min Error: {errors.min()}")
print(f"Mean Error: {errors.mean()}")
print(f"Total Error: {errors.sum()}")
```

# 🟡 LEVEL 2: The Structural Engineer

**Goal:** Master dimensions. Shape mismatches are the #1 source of ML bugs.

## 2.1 The Shape Trio

```python
img = np.zeros((28, 28, 3))  # Height, Width, Channels

print(f"Shape: {img.shape}")  # (28, 28, 3)
print(f"Rank (ndim): {img.ndim}") # 3 dimensions
print(f"Size: {img.size}")    # 2352 total pixels
```

## 2.2 Reshaping (The Magic -1)

Often need to flatten images for dense layers.

```python
flat = np.arange(12)
print(f"Original: {flat}")

# Reshape to matrix
matrix = flat.reshape(3, 4)
print(f"\n(3, 4) Matrix:\n{matrix}")

# Automatic calculation (-1)
# "2 rows, and however many columns are needed"
auto = flat.reshape(2, -1)
print(f"\n(2, -1) Matrix:\n{auto}")
```

## 2.3 Indexing & Slicing

`arr[row, col]` syntax.

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(f"Row 0: {data[0]}")            # [10, 20, 30]
print(f"Col 2: {data[:, 2]}")        # [30, 60, 90]
print(f"Sub-grid:\n{data[:2, :2]}") # Top-left 2x2
```

## 2.4 Axis Logic (CRITICAL)

- `axis=0`: Collapse **rows** (move down ↓). Compute statistics "per column".
- `axis=1`: Collapse **cols** (move across →). Compute statistics "per row".

```python
batch = np.array([
    [0.9, 0.1],  # Prediction 1
    [0.6, 0.4],  # Prediction 2
    [0.2, 0.8]   # Prediction 3
])

# Mean probability per CLASS (down columns)
class_means = batch.mean(axis=0)
print(f"Class Means (axis=0): {class_means}")

# Max probability per SAMPLE (across rows)
best_probs = batch.max(axis=1)
print(f"Best Probs (axis=1): {best_probs}")
```

# 🔴 LEVEL 3: The Advanced Computer

**Goal:** Write vectorized code that looks like math equations.

## 3.1 Broadcasting

NumPy's superpower. It stretches smaller arrays to match shape.

```python
X = np.array([[10, 20, 30], 
              [40, 50, 60]]) 

mean = np.array([10, 20, 30])

# Shapes: (2, 3) - (3,)
# NumPy stretches (3,) to (2, 3) automatically!
centered = X - mean
print(f"Centerd X:\n{centered}")
```

## 3.2 Boolean Masking

Filter data without `if` statements.

```python
scores = np.array([0.1, 0.9, 0.8, 0.4])

# Create mask
mask = scores > 0.5
print(f"Mask: {mask}")

# Filter
passing = scores[mask]
print(f"Passing scores: {passing}")

# ML Trick: ReLU Activation (x > 0 ? x : 0)
relu = np.maximum(scores - 0.5, 0)
print(f"ReLU output: {relu}")
```

## 3.3 Matrix Multiplication (@)

The core of every neural network layer: $Y = XW + b$

```python
X = np.random.rand(32, 128)  # Batch=32, Features=128
W = np.random.rand(128, 64)  # Input=128, Output=64
b = np.random.rand(64)       # Bias

# Forward pass
Y = X @ W + b
print(f"Output Shape: {Y.shape}") # Should be (32, 64)
```

## 4.2 Stacking & Concatenating

Combining batches of data.

```python
img1 = np.zeros((10, 10))
img2 = np.ones((10, 10))

# Stack: Create NEW dimension (Batch)
batch = np.stack([img1, img2])
print(f"Batch Shape: {batch.shape}") # (2, 10, 10)

# Concatenate: Join along EXISTING dimension
wide_img = np.concatenate([img1, img2], axis=1)
print(f"Wide Image Shape: {wide_img.shape}")
```

## 4.3 Dimension Surgery (`expand_dims`, `squeeze`)

Models often expect `(Batch, H, W, C)` even if you only have 1 image.

```python
img = np.zeros((28, 28))  # Single grayscale image

# Add Batch dim
batch_img = np.expand_dims(img, axis=0)
print(f"Batch Dim: {batch_img.shape}")

# Add Channel dim
cnn_input = np.expand_dims(batch_img, axis=-1)
print(f"CNN Input: {cnn_input.shape}")

# Remove singleton dims
flat = cnn_input.squeeze()
print(f"Squeezed: {flat.shape}")
```

### 🏆 Final Challenge: Image Preprocessing Pipeline

Combine **broadcasting**, **reshaping**, **stacking**, and **normalization** in one function.

```python
def preprocess_images(img_list):
    # 1. Stack list into array
    batch = np.stack(img_list)
    
    # 2. Normalize to [0, 1]
    norm = batch.astype(np.float32) / 255.0
    
    # 3. Add Channel dim if missing (assume Grayscale)
    if norm.ndim == 3: # (B, H, W)
        norm = np.expand_dims(norm, axis=-1)
        
    return norm

# Simulating 3 images of 32x32
raw_images = [np.random.randint(0, 255, (32, 32)) for _ in range(3)]
processed = preprocess_images(raw_images)

print(f"Final Input Shape: {processed.shape}")
print(f"Value Range: {processed.min():.2f} - {processed.max():.2f}")
```
