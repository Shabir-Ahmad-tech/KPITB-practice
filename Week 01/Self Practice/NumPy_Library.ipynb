# 🧠 Module 2: The "Brain" of ML - Broadcasting, Reshaping & Axes

**This is the MOST IMPORTANT module.** Most bugs in ML code come from shape mismatches.

## 🎯 Top 3 Functions to Memorize

| Function | ML Use Case |
|----------|-------------|
| `.reshape(-1, 1)` | Convert 1D labels to column vector |
| `np.expand_dims()` | Add batch dimension for model input |
| `axis=0` vs `axis=1` | Sum over samples vs features |

---

### 2.1 Understanding Axes

**The Golden Rule:** `axis=0` operates DOWN (over rows/samples), `axis=1` operates ACROSS (over columns/features).

```python
# Dataset: 4 samples, 3 features
X = np.array([
    [1, 2, 3],    # Sample 0
    [4, 5, 6],    # Sample 1
    [7, 8, 9],    # Sample 2
    [10, 11, 12]  # Sample 3
])
print(f"Shape: {X.shape}")
```

```python
# axis=0: Collapse rows → Result has shape (3,)
mean_per_feature = X.mean(axis=0)  # Mean of each feature across all samples
print(f"Mean per feature: {mean_per_feature}")

# axis=1: Collapse columns → Result has shape (4,)
mean_per_sample = X.mean(axis=1)   # Mean of each sample across all features
print(f"Mean per sample: {mean_per_sample}")
```

### 2.2 Reshaping with -1 (Auto-Calculate)

**Why -1 is magic:** NumPy auto-calculates that dimension. Essential for flattening images.

```python
# Simulated grayscale image: 28x28 pixels
image = np.random.rand(28, 28)
print(f"Original shape: {image.shape}")

# Flatten for Dense layer: (784,)
flattened = image.reshape(-1)  # -1 means "figure it out"
print(f"Flattened: {flattened.shape}")

# Add batch dimension: (1, 784)
batched = image.reshape(1, -1)
print(f"With batch: {batched.shape}")
```

```python
# Convert 1D labels to column vector (required for many operations)
labels = np.array([0, 1, 1, 0, 2])
print(f"Original: {labels.shape}")

labels_col = labels.reshape(-1, 1)  # Shape: (5, 1)
print(f"Column vector:\n{labels_col}")
```

### 2.3 expand_dims vs reshape vs None indexing

**Why we need this:** Neural networks expect batch dimensions. A single image (28, 28) must become (1, 28, 28).

```python
# Single image
img = np.random.rand(28, 28)
print(f"Original: {img.shape}")

# Method 1: np.expand_dims (most readable)
batched_1 = np.expand_dims(img, axis=0)
print(f"expand_dims: {batched_1.shape}")

# Method 2: None indexing (shortest)
batched_2 = img[None, :, :]
print(f"None indexing: {batched_2.shape}")

# Method 3: reshape
batched_3 = img.reshape(1, 28, 28)
print(f"Reshape: {batched_3.shape}")
```

### 2.4 Broadcasting: The Secret Weapon

**Broadcasting Rules:**
1. Compare shapes from RIGHT to LEFT
2. Dimensions must be equal OR one of them is 1
3. Missing dimensions are treated as 1

```python
# Example 1: Normalize pixel values (0-255 → 0-1)
pixels = np.array([[100, 150, 200], [50, 100, 150]], dtype=np.float32)
normalized = pixels / 255.0  # Scalar broadcasts to all elements
print(f"Normalized:\n{normalized}")
```

```python
# Example 2: Subtract mean per feature (StandardScaler logic)
X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
], dtype=np.float32)

mean = X.mean(axis=0)  # Shape: (3,)
print(f"X shape: {X.shape}, Mean shape: {mean.shape}")

# Broadcasting: (3,3) - (3,) → (3,3) - (1,3) → Works!
X_centered = X - mean
print(f"Centered:\n{X_centered}")
```

```python
# Example 3: RGB channel normalization (ImageNet preprocessing)
# Image shape: (H, W, C) = (224, 224, 3)
image = np.random.randint(0, 256, (224, 224, 3)).astype(np.float32)

# ImageNet means for R, G, B channels
channel_means = np.array([123.68, 116.779, 103.939])

# Broadcasting: (224, 224, 3) - (3,) → Works!
normalized = image - channel_means
print(f"Image shape: {image.shape}")
print(f"After normalization - Min: {normalized.min():.1f}, Max: {normalized.max():.1f}")
```

### 3.1 Boolean Masking

### 3.2 np.where - Conditional Selection

```python
# Replace negative predictions with 0 (ReLU behavior)
predictions = np.array([-0.5, 1.2, -0.1, 0.8, -2.0, 3.5])
relu_output = np.where(predictions > 0, predictions, 0)
print(f"ReLU output: {relu_output}")
```

```python
# Get indices where condition is true
indices = np.where(predictions > 0)[0]
print(f"Positive indices: {indices}")
```

### 3.3 argmax/argmin - Critical for Classification

```python
# Softmax output from a 3-class classifier
probabilities = np.array([
    [0.1, 0.7, 0.2],  # Sample 0: Class 1
    [0.8, 0.1, 0.1],  # Sample 1: Class 0
    [0.2, 0.3, 0.5],  # Sample 2: Class 2
])

# Get predicted classes
predictions = np.argmax(probabilities, axis=1)
print(f"Predicted classes: {predictions}")
```

### 4.1 Matrix Multiplication

### 4.2 Random Sampling for Initialization

```python
# Xavier initialization for sigmoid/tanh activations
fan_in, fan_out = 784, 256
xavier_weights = np.random.randn(fan_in, fan_out) * np.sqrt(2.0 / (fan_in + fan_out))
print(f"Xavier init - Mean: {xavier_weights.mean():.4f}, Std: {xavier_weights.std():.4f}")

# He initialization for ReLU
he_weights = np.random.randn(fan_in, fan_out) * np.sqrt(2.0 / fan_in)
print(f"He init - Mean: {he_weights.mean():.4f}, Std: {he_weights.std():.4f}")
```

### 4.3 Softmax Implementation

```python
def softmax(x):
    """Numerically stable softmax."""
    exp_x = np.exp(x - x.max(axis=-1, keepdims=True))  # Subtract max for stability
    return exp_x / exp_x.sum(axis=-1, keepdims=True)

logits = np.array([[2.0, 1.0, 0.1], [1.0, 3.0, 0.5]])
probs = softmax(logits)
print(f"Probabilities:\n{probs}")
print(f"Sum per sample: {probs.sum(axis=1)}")
```

```python
# =============================================
# 🏆 COMPLETE SOLUTION
# =============================================

np.random.seed(42)

# Step 1: Create dataset (given)
images = np.random.randint(0, 256, (10000, 64, 64, 3), dtype=np.uint8)
corrupt_indices = np.random.choice(10000, 50, replace=False)
images[corrupt_indices] = 0

print("=" * 50)
print("STEP 1: Original Data")
print(f"Shape: {images.shape}, Dtype: {images.dtype}")
```

```python
# Step 2: Remove corrupted images
# Sum all pixels per image, then check if > 0
pixel_sums = images.sum(axis=(1, 2, 3))  # Sum over H, W, C
clean_mask = pixel_sums > 0
clean_images = images[clean_mask]

print("\nSTEP 2: Removed Corrupted Images")
print(f"Removed: {10000 - len(clean_images)} images")
print(f"Clean shape: {clean_images.shape}")
```

```python
# Step 3: Normalize to [0, 1]
normalized = clean_images.astype(np.float32) / 255.0

print("\nSTEP 3: Normalized to [0, 1]")
print(f"Min: {normalized.min():.4f}, Max: {normalized.max():.4f}")
```

```python
# Step 4: Subtract ImageNet means (per channel)
imagenet_mean = np.array([123.68, 116.779, 103.939]) / 255.0  # Scale to 0-1
standardized = normalized - imagenet_mean  # Broadcasting: (N,64,64,3) - (3,)

print("\nSTEP 4: ImageNet Standardization")
print(f"Mean per channel: {standardized.mean(axis=(0,1,2)).round(4)}")
```

```python
# Step 5: Random horizontal flip (50% of images)
flip_mask = np.random.rand(len(standardized)) < 0.5
augmented = standardized.copy()
augmented[flip_mask] = np.flip(augmented[flip_mask], axis=2)  # Flip on width axis

print("\nSTEP 5: Data Augmentation (Horizontal Flip)")
print(f"Flipped images: {flip_mask.sum()}")
```

```python
# Step 6: Flatten for Dense layer
flattened = augmented.reshape(len(augmented), -1)

print("\nSTEP 6: Flattened for Dense Layer")
print(f"Shape: {flattened.shape}")
```

```python
# Step 7: Train/Val split (80/20)
n_samples = len(flattened)
n_train = int(0.8 * n_samples)

# Shuffle indices
indices = np.arange(n_samples)
np.random.shuffle(indices)

train_idx, val_idx = indices[:n_train], indices[n_train:]
X_train, X_val = flattened[train_idx], flattened[val_idx]

print("\nSTEP 7: Train/Val Split")
print(f"Training set: {X_train.shape}")
print(f"Validation set: {X_val.shape}")
```

```python
# 🏆 FINAL SUMMARY
print("\n" + "=" * 50)
print("🏆 PIPELINE COMPLETE!")
print("=" * 50)
print(f"Original: {10000} images")
print(f"After cleaning: {len(clean_images)} images")
print(f"Training: {len(X_train)} samples")
print(f"Validation: {len(X_val)} samples")
print(f"Feature dimension: {X_train.shape[1]}")
print(f"Memory (training): {X_train.nbytes / 1024**2:.2f} MB")
```
