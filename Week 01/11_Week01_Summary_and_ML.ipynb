# 🤖 PART 2: Mini ML Pipeline

Let's build a simple machine learning pipeline using ONLY what we learned this week!

## Step 1: Create Sample Dataset 📊

```python
print("--- Step 1: Dataset Creation ---")

# Simulated Iris-like data (features for flowers)
# Each sample: [sepal_length, sepal_width, petal_length, petal_width]
dataset = [
    ([5.1, 3.5, 1.4, 0.2], "setosa"),
    ([4.9, 3.0, 1.4, 0.2], "setosa"),
    ([7.0, 3.2, 4.7, 1.4], "versicolor"),
    ([6.4, 3.2, 4.5, 1.5], "versicolor"),
    ([6.3, 3.3, 6.0, 2.5], "virginica"),
    ([5.8, 2.7, 5.1, 1.9], "virginica"),
    ([5.0, 3.4, 1.5, 0.2], "setosa"),
    ([6.5, 2.8, 4.6, 1.5], "versicolor"),
    ([6.5, 3.0, 5.8, 2.2], "virginica"),
    ([4.8, 3.4, 1.6, 0.2], "setosa"),
]

print(f"Dataset size: {len(dataset)} samples")
print(f"Sample: Features={dataset[0][0]}, Label={dataset[0][1]}")
```

## Step 2: Data Preprocessing Functions 🔧

```python
print("--- Step 2: Preprocessing Functions ---")

def split_data(data, train_ratio=0.8):
    """
    Split dataset into training and testing sets.
    
    Args:
        data: List of (features, label) tuples
        train_ratio: Percentage for training
        
    Returns:
        Tuple of (train_data, test_data)
    """
    split_idx = int(len(data) * train_ratio)
    train = data[:split_idx]
    test = data[split_idx:]
    return train, test

def get_unique_labels(data):
    """Extract unique labels from dataset."""
    labels = [sample[1] for sample in data]
    return set(labels)

def normalize_features(features, min_vals, max_vals):
    """Normalize features to 0-1 range."""
    normalized = []
    for i, val in enumerate(features):
        if max_vals[i] - min_vals[i] != 0:
            norm_val = (val - min_vals[i]) / (max_vals[i] - min_vals[i])
        else:
            norm_val = 0
        normalized.append(norm_val)
    return normalized

# Apply split
train_data, test_data = split_data(dataset, 0.8)
print(f"Training samples: {len(train_data)}")
print(f"Testing samples: {len(test_data)}")
print(f"Unique labels: {get_unique_labels(dataset)}")
```

## Step 3: Simple K-Nearest Neighbor (From Scratch!) 🎯

```python
print("--- Step 3: Simple KNN Classifier ---")

def euclidean_distance(point1, point2):
    """Calculate Euclidean distance between two points."""
    total = 0
    for i in range(len(point1)):
        total += (point1[i] - point2[i]) ** 2
    return total ** 0.5

def knn_predict(train_data, test_point, k=3):
    """Predict label using K-Nearest Neighbors."""
    # Calculate distances to all training points
    distances = []
    for train_features, train_label in train_data:
        dist = euclidean_distance(test_point, train_features)
        distances.append((dist, train_label))
    
    # Sort by distance and get k nearest
    distances.sort()
    k_nearest = distances[:k]
    
    # Vote for most common label
    label_counts = {}
    for _, label in k_nearest:
        label_counts[label] = label_counts.get(label, 0) + 1
    
    # Return label with most votes
    best_label = None
    best_count = 0
    for label, count in label_counts.items():
        if count > best_count:
            best_count = count
            best_label = label
    
    return best_label

# Test on one sample
test_features = test_data[0][0]
test_label = test_data[0][1]
prediction = knn_predict(train_data, test_features, k=3)

print(f"Test features: {test_features}")
print(f"Actual label: {test_label}")
print(f"Predicted: {prediction}")
print(f"Correct: {'✓' if prediction == test_label else '✗'}")
```

## Step 4: Evaluate on All Test Data 📈

```python
print("--- Step 4: Full Evaluation ---")

def evaluate_model(train_data, test_data, k=3):
    """Evaluate model on test data."""
    correct = 0
    predictions = []
    actuals = []
    
    for features, actual_label in test_data:
        pred = knn_predict(train_data, features, k)
        predictions.append(pred)
        actuals.append(actual_label)
        
        if pred == actual_label:
            correct += 1
    
    accuracy = correct / len(test_data)
    return accuracy, predictions, actuals

# Evaluate
accuracy, preds, actuals = evaluate_model(train_data, test_data, k=3)

print("Predictions vs Actuals:")
for i, (pred, actual) in enumerate(zip(preds, actuals)):
    status = "✓" if pred == actual else "✗"
    print(f"  Sample {i+1}: Pred={pred:12} | Actual={actual:12} | {status}")

print(f"\nAccuracy: {accuracy:.2%}")
```

## Step 5: Experiment with Different K Values 🔬

```python
print("--- Step 5: Hyperparameter Tuning ---")

k_values = [1, 3, 5]
results = {}

print("Testing different K values:")
print("-" * 30)

for k in k_values:
    acc, _, _ = evaluate_model(train_data, test_data, k=k)
    results[k] = acc
    print(f"  K={k}: Accuracy = {acc:.2%}")

# Find best K
best_k = None
best_acc = 0
for k, acc in results.items():
    if acc > best_acc:
        best_acc = acc
        best_k = k

print("-" * 30)
print(f"Best K: {best_k} with accuracy {best_acc:.2%}")
```

# 🏆 PART 3: Complete ML Pipeline Function

```python
print("=" * 50)
print("COMPLETE ML PIPELINE")
print("=" * 50)

def run_ml_pipeline(data, config):
    """
    Run complete ML pipeline.
    
    Args:
        data: List of (features, label) tuples
        config: Dictionary with 'train_ratio' and 'k'
        
    Returns:
        Dictionary with results
    """
    print("\n[1] Splitting data...")
    train, test = split_data(data, config["train_ratio"])
    print(f"    Train: {len(train)}, Test: {len(test)}")
    
    print("\n[2] Analyzing labels...")
    labels = get_unique_labels(data)
    print(f"    Classes: {labels}")
    
    print("\n[3] Training & Evaluating...")
    accuracy, preds, actuals = evaluate_model(train, test, config["k"])
    print(f"    K = {config['k']}")
    print(f"    Accuracy = {accuracy:.2%}")
    
    results = {
        "accuracy": accuracy,
        "num_train": len(train),
        "num_test": len(test),
        "num_classes": len(labels),
        "k": config["k"],
        "predictions": preds
    }
    
    return results

# Run the pipeline
pipeline_config = {
    "train_ratio": 0.8,
    "k": 3
}

results = run_ml_pipeline(dataset, pipeline_config)

print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)
for key, value in results.items():
    if key != "predictions":
        print(f"  {key}: {value}")
```
