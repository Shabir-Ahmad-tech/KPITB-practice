## 📖 What is a For Loop?
A **for loop** repeats code for each item in a sequence. Think of it as attacking each enemy in a group one by one!

**Syntax:**
```python
for item in sequence:
    # do something with item
```

---
## 📖 1. Basic For Loop

````python
print("--- 1. Basic For Loop ---")

# Loop through a list
enemies = ["Goblin", "Orc", "Troll"]

for enemy in enemies:
    print(f"You attacked {enemy}!")

print("\nAll enemies defeated!")
````

## 📖 2. Looping with range()

````python
print("--- 2. range() Function ---")

# range(stop) - 0 to stop-1
print("range(5):")
for i in range(5):
    print(f"  {i}")

# range(start, stop)
print("\nrange(2, 6):")
for i in range(2, 6):
    print(f"  {i}")

# range(start, stop, step)
print("\nrange(0, 10, 2):")
for i in range(0, 10, 2):
    print(f"  {i}")
````

### 🤖 AI/ML Application: Epoch Training Loop

````python
# Simulating training epochs
num_epochs = 5
initial_loss = 1.0

print("Training Progress:")
print("-" * 30)

loss = initial_loss
for epoch in range(1, num_epochs + 1):
    # Simulate loss decrease
    loss = loss * 0.7
    accuracy = 1 - loss
    print(f"Epoch {epoch}/{num_epochs} - Loss: {loss:.4f}, Acc: {accuracy:.2%}")

print("-" * 30)
print("Training Complete!")
````

## 📖 4. enumerate() - Get Index and Value

````python
print("--- 4. enumerate() ---")

inventory = ["Sword", "Shield", "Potion", "Key"]

# Without enumerate (manual index)
print("Manual indexing:")
for i in range(len(inventory)):
    print(f"  {i}: {inventory[i]}")

# With enumerate (cleaner!)
print("\nWith enumerate:")
for index, item in enumerate(inventory):
    print(f"  {index}: {item}")

# Start from 1
print("\nStarting from 1:")
for index, item in enumerate(inventory, start=1):
    print(f"  {index}. {item}")
````

### 🤖 AI/ML Application: Batch Processing

````python
# Processing data in batches
all_data = list(range(100))  # 100 samples
batch_size = 32

print(f"Total samples: {len(all_data)}")
print(f"Batch size: {batch_size}")
print("-" * 30)

for batch_num, start_idx in enumerate(range(0, len(all_data), batch_size)):
    end_idx = min(start_idx + batch_size, len(all_data))
    batch = all_data[start_idx:end_idx]
    print(f"Batch {batch_num + 1}: samples {start_idx}-{end_idx-1} ({len(batch)} items)")
````

## 📖 5. zip() - Loop Over Multiple Lists

````python
print("--- 5. zip() ---")

players = ["Ali", "Hamza", "Zain"]
scores = [100, 85, 92]

# Loop through both lists together
print("Leaderboard:")
for player, score in zip(players, scores):
    print(f"  {player}: {score} points")

# With enumerate for ranking
print("\nRanked Leaderboard:")
for rank, (player, score) in enumerate(zip(players, scores), start=1):
    print(f"  #{rank} {player}: {score}")
````

### 🤖 AI/ML Application: Predictions vs Actuals

````python
# Comparing predictions to actual values
actual = [1, 0, 1, 1, 0, 1]
predicted = [1, 0, 0, 1, 0, 1]

correct = 0
total = len(actual)

print("Sample | Actual | Predicted | Correct")
print("-" * 40)

for i, (act, pred) in enumerate(zip(actual, predicted)):
    is_correct = "✓" if act == pred else "✗"
    if act == pred:
        correct += 1
    print(f"   {i}   |   {act}    |     {pred}     |    {is_correct}")

print("-" * 40)
print(f"Accuracy: {correct}/{total} = {correct/total:.2%}")
````

## 📖 6. Nested Loops

````python
print("--- 6. Nested Loops ---")

# Create a 3x3 game grid
print("Game Grid:")
for row in range(3):
    for col in range(3):
        print(f"({row},{col})", end=" ")
    print()  # New line after each row

# Multiplication table
print("\nMultiplication Table (1-3):")
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i*j:3}", end=" ")
    print()
````

## 📖 7. List Comprehensions (Pythonic!)

````python
print("--- 7. List Comprehensions ---")

# Traditional loop
squares = []
for x in range(5):
    squares.append(x ** 2)
print(f"Traditional: {squares}")

# List comprehension (same result, one line!)
squares = [x ** 2 for x in range(5)]
print(f"Comprehension: {squares}")

# With condition
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print(f"Even squares: {even_squares}")
````

### 🤖 AI/ML Application: Data Transformation

````python
# Normalize values to 0-1 range
raw_data = [10, 20, 30, 40, 50]
min_val = min(raw_data)
max_val = max(raw_data)

normalized = [(x - min_val) / (max_val - min_val) for x in raw_data]
print(f"Raw: {raw_data}")
print(f"Normalized: {normalized}")

# Convert probabilities to binary predictions
probabilities = [0.3, 0.7, 0.5, 0.9, 0.2]
threshold = 0.5
predictions = [1 if p >= threshold else 0 for p in probabilities]
print(f"\nProbabilities: {probabilities}")
print(f"Predictions: {predictions}")
````

# 📝 Key Takeaways

| Concept | Usage |
|---------|-------|
| `for x in list` | Iterate over items |
| `range(n)` | Generate 0 to n-1 |
| `enumerate()` | Get index + value |
| `zip()` | Loop multiple lists |
| List comprehension | `[expr for x in list]` |

**ML Connections:**
- Training epochs: `for epoch in range(n)`
- Batch processing: `range(0, len(data), batch_size)`
- Comparing predictions: `zip(actual, predicted)`
- Data transformation: List comprehensions
