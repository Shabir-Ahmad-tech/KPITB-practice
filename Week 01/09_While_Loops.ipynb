## 📖 While vs For Loops

| For Loop | While Loop |
|----------|------------|
| Known iterations | Unknown iterations |
| Iterate over sequence | Continue until condition |
| `for i in range(n)` | `while condition:` |

---
## 📖 1. Basic While Loop

````python
print("--- 1. Basic While Loop ---")

# Continue while condition is True
hp = 100
damage = 30

print(f"Starting HP: {hp}")
while hp > 0:
    hp -= damage
    print(f"  Hit! HP now: {max(hp, 0)}")

print("Enemy defeated!")
````

---
## 📖 2. Counter-Based While

````python
print("--- 2. Counter-Based While ---")

# Countdown
count = 5
print("Launch in:")
while count > 0:
    print(f"  {count}...")
    count -= 1
print("🚀 Liftoff!")

# Count up
level = 1
max_level = 5
print("\nLeveling up:")
while level <= max_level:
    print(f"  Level {level}")
    level += 1
````

### 🤖 AI/ML Application: Training Until Convergence

````python
# Train until loss is below threshold
loss = 1.0
target_loss = 0.1
learning_rate = 0.3
epoch = 0
max_epochs = 100  # Safety limit!

print("Training until convergence:")
while loss > target_loss and epoch < max_epochs:
    epoch += 1
    loss = loss * (1 - learning_rate)  # Simulate loss decrease
    print(f"  Epoch {epoch}: Loss = {loss:.4f}")

if loss <= target_loss:
    print(f"\n✓ Converged at epoch {epoch}!")
else:
    print(f"\n⚠️ Max epochs reached")
````

## 📖 3. Avoiding Infinite Loops

````python
print("--- 3. Avoiding Infinite Loops ---")

# DANGER: This would run forever!
# while True:
#     print("Infinite!")

# SAFE: Always have an exit condition
count = 0
max_iterations = 5

while True:  # Potentially infinite
    print(f"  Iteration {count}")
    count += 1
    
    if count >= max_iterations:
        print("  Exiting safely!")
        break  # ESCAPE!
````

## 📖 4. While with User Input (Simulated)

````python
print("--- 4. Input-Based While ---")

# Simulated user inputs
user_inputs = ["attack", "defend", "heal", "run", "quit"]
input_index = 0

def get_input():
    global input_index
    if input_index < len(user_inputs):
        result = user_inputs[input_index]
        input_index += 1
        return result
    return "quit"

print("Game started! (simulated inputs)")
command = get_input()

while command != "quit":
    print(f"  > {command}")
    command = get_input()

print("Game ended.")
````

## 📖 5. While-Else

````python
print("--- 5. While-Else ---")

# Else runs if while completes normally (no break)
attempts = 3
password = "secret"
guesses = ["wrong1", "wrong2", "secret"]
guess_index = 0

while attempts > 0:
    guess = guesses[guess_index]
    guess_index += 1
    print(f"  Attempt: {guess}")
    
    if guess == password:
        print("  ✓ Access granted!")
        break
    
    attempts -= 1
else:
    print("  ✗ Access denied! No more attempts.")
````

### 🤖 AI/ML Application: Hyperparameter Search

````python
# Search for optimal learning rate
lr = 0.0001
max_lr = 1.0
multiplier = 2
best_accuracy = 0
best_lr = lr

# Simulated accuracies for different learning rates
def simulate_training(learning_rate):
    # Peak accuracy around 0.01
    if learning_rate < 0.001:
        return 0.7
    elif learning_rate < 0.1:
        return 0.92
    else:
        return 0.6  # Too high, unstable

print("Learning Rate Search:")
while lr < max_lr:
    accuracy = simulate_training(lr)
    print(f"  LR={lr:.4f}: Accuracy={accuracy:.2f}")
    
    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_lr = lr
    
    lr *= multiplier

print(f"\nBest LR: {best_lr} with accuracy {best_accuracy:.2f}")
````

# 📝 Key Takeaways

| Concept | Syntax |
|---------|--------|
| Basic while | `while condition:` |
| With break | `while True: ... break` |
| Counter | `while count < max:` |
| While-else | `while: ... else:` |

**ML Connections:**
- Training until convergence: `while loss > target`
- Hyperparameter search: `while lr < max_lr`
- Early stopping with patience
