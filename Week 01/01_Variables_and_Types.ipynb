## 📖 1. Variables & Naming

**What is a Variable?**
A variable is like a labeled box where you store data. In Python, you don't need to declare the type - Python figures it out automatically!

**Naming Rules:**
- Must start with a letter or underscore `_`
- Can contain letters, numbers, and underscores
- Case-sensitive (`name` ≠ `Name`)
- No spaces! Use `snake_case` (e.g., `player_name`)

````python
print("--- 1. Variables & Naming ---")

# Creating a variable is simple: name = value
player_name = "Shadow_Stalker"
print("Player Name:", player_name)

# You can change a variable's value anytime
player_name = "Dragon_Knight"
print("New Name:", player_name)
````

## 📖 2. Core Data Types

Python has 4 fundamental data types:

| Type | Description | Example | ML Use Case |
|------|-------------|---------|-------------|
| `str` | Text/String | `"hello"` | Labels, categories |
| `int` | Whole number | `42` | Counts, indices |
| `float` | Decimal number | `3.14` | Weights, probabilities |
| `bool` | True/False | `True` | Flags, conditions |

````python
print("--- 2. Core Data Types ---")

# A. STRINGS (Text) - Use quotes
character_class = "Warrior"
print(f"Class: {character_class} (Type: {type(character_class).__name__})")

# B. INTEGERS (Whole Numbers)
health_points = 100
level = 15
print(f"HP: {health_points}, Level: {level} (Type: {type(health_points).__name__})")

# C. FLOATS (Decimal Numbers)
movement_speed = 5.5
print(f"Speed: {movement_speed} (Type: {type(movement_speed).__name__})")

# D. BOOLEANS (True or False)
is_online = True
print(f"Online: {is_online} (Type: {type(is_online).__name__})")
````

## 📖 3. Math Operators

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division | `5 / 3` | `1.666...` |
| `//` | Floor Division | `5 // 3` | `1` |
| `%` | Modulus (Remainder) | `5 % 3` | `2` |
| `**` | Power | `5 ** 3` | `125` |

````python
print("--- 3. Math Operators ---")
a = 10
b = 3

print(f"Addition (10 + 3): {a + b}")
print(f"Subtraction (10 - 3): {a - b}")
print(f"Multiplication (10 * 3): {a * b}")
print(f"Division (10 / 3): {a / b:.2f}")  # .2f = 2 decimal places
print(f"Floor Division (10 // 3): {a // b}")  # Removes decimal
print(f"Modulus (10 % 3): {a % b}")  # Remainder
print(f"Power (10 ** 3): {a ** b}")  # 10 to the power of 3
````

### 🤖 AI/ML Application: Math in Machine Learning

````python
# Example: Calculating Mean Squared Error (MSE) for 1 prediction
actual_value = 100
predicted_value = 95

error = actual_value - predicted_value
squared_error = error ** 2

print("MSE Calculation:")
print(f"  Actual: {actual_value}, Predicted: {predicted_value}")
print(f"  Error: {error}")
print(f"  Squared Error: {squared_error}")

# Example: Normalizing a value to 0-1 range
value = 75
min_val = 0
max_val = 100
normalized = (value - min_val) / (max_val - min_val)
print(f"\nNormalized {value} to range [0,1]: {normalized}")
````

## 📖 4. Assignment Operators (Shortcuts)

Instead of writing `x = x + 5`, use the shortcut `x += 5`

| Operator | Equivalent | Example |
|----------|------------|--------|
| `+=` | `x = x + n` | `x += 5` |
| `-=` | `x = x - n` | `x -= 5` |
| `*=` | `x = x * n` | `x *= 2` |
| `/=` | `x = x / n` | `x /= 2` |

````python
print("--- 4. Assignment Operators ---")

coins = 100
print(f"Starting coins: {coins}")

coins += 50  # Same as coins = coins + 50
print(f"After += 50: {coins}")

coins -= 20  # Same as coins = coins - 20
print(f"After -= 20: {coins}")

coins *= 2   # Double the coins
print(f"After *= 2: {coins}")

coins //= 3  # Integer divide by 3
print(f"After //= 3: {coins}")
````

## 📖 5. Comparison Operators

These operators compare values and return `True` or `False`.

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `5 == 5` → `True` |
| `!=` | Not equal | `5 != 3` → `True` |
| `>` | Greater than | `5 > 3` → `True` |
| `<` | Less than | `5 < 3` → `False` |
| `>=` | Greater or equal | `5 >= 5` → `True` |
| `<=` | Less or equal | `5 <= 3` → `False` |

````python
print("--- 5. Comparison Operators ---")

my_hp = 50
enemy_hp = 50

print(f"my_hp == enemy_hp: {my_hp == enemy_hp}")
print(f"my_hp != 100: {my_hp != 100}")
print(f"my_hp > 20: {my_hp > 20}")
print(f"my_hp < 100: {my_hp < 100}")
print(f"my_hp >= 50: {my_hp >= 50}")
print(f"my_hp <= 25: {my_hp <= 25}")
````

### 🤖 AI/ML Application: Thresholds and Comparisons

````python
# Classification thresholds
prediction_probability = 0.75
threshold = 0.5

is_positive_class = prediction_probability >= threshold
print(f"Probability: {prediction_probability}")
print(f"Threshold: {threshold}")
print(f"Classify as Positive? {is_positive_class}")

# Accuracy check
accuracy = 0.92
target_accuracy = 0.90
model_is_good = accuracy >= target_accuracy
print(f"\nModel accuracy ({accuracy}) meets target ({target_accuracy})? {model_is_good}")
````

## 📖 6. Logical Operators

Combine multiple conditions:

| Operator | Description | Example |
|----------|-------------|--------|
| `and` | Both must be True | `True and False` → `False` |
| `or` | At least one True | `True or False` → `True` |
| `not` | Flip the value | `not True` → `False` |

````python
print("--- 6. Logical Operators ---")

has_key = True
has_energy = False

# AND - Both conditions must be True
can_open_door = has_key and has_energy
print(f"Key: {has_key}, Energy: {has_energy}")
print(f"Can open door (AND): {can_open_door}")

# OR - At least one condition must be True
can_try = has_key or has_energy
print(f"Can try (OR): {can_try}")

# NOT - Flips the value
is_locked = not has_key
print(f"Is locked (NOT has_key): {is_locked}")
````

### 🤖 AI/ML Application: Data Validation

````python
# Checking if data is valid for training
sample_count = 1000
has_labels = True
no_missing_values = True

min_samples = 100
enough_data = sample_count >= min_samples

# All conditions must be met
ready_to_train = enough_data and has_labels and no_missing_values

print(f"Samples: {sample_count} (min: {min_samples})")
print(f"Has labels: {has_labels}")
print(f"No missing values: {no_missing_values}")
print(f"Ready to train: {ready_to_train}")
````

## 📖 7. Type Conversion (Casting)

Convert between types using these functions:

| Function | Converts to | Example |
|----------|-------------|--------|
| `str()` | String | `str(42)` → `"42"` |
| `int()` | Integer | `int("42")` → `42` |
| `float()` | Float | `float("3.14")` → `3.14` |
| `bool()` | Boolean | `bool(1)` → `True` |

````python
print("--- 7. Type Conversion ---")

# int to string (for concatenation)
score = 500
message = "Your score is: " + str(score)
print(message)

# string to int (for math)
xp_input = "250"
total_xp = int(xp_input) + 50
print(f"Total XP: {total_xp}")

# float to int (removes decimal)
price = 19.99
rounded_price = int(price)
print(f"Price: {price} → Rounded: {rounded_price}")

# Truthiness - what converts to True/False?
print(f"\nbool(0): {bool(0)}")
print(f"bool(1): {bool(1)}")
print(f"bool(''): {bool('')}")
print(f"bool('hello'): {bool('hello')}")
````
