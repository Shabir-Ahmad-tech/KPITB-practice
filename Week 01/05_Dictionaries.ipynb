## 📖 What is a Dictionary?
A **Dictionary** is a collection of **key-value pairs**. Think of it as a spell book where each spell name (key) has a description (value)!

**Key Properties:**
- Unordered (Python 3.7+ maintains insertion order)
- Keys must be unique and immutable
- Values can be any type
- Uses curly braces `{key: value}`
- Very fast lookup by key

---
## 📖 1. Creating Dictionaries

````python
print("--- 1. Creating Dictionaries ---")

# Curly braces with key: value pairs
player = {
    "name": "Shadow_Knight",
    "level": 25,
    "health": 100.0,
    "is_online": True
}
print("Player:", player)

# Empty dictionary
empty_dict = {}
print("Empty dict:", empty_dict)

# Using dict() constructor
enemy = dict(name="Goblin", health=50, damage=10)
print("Enemy:", enemy)

# From list of tuples
items = dict([("sword", 100), ("shield", 75), ("potion", 25)])
print("Items:", items)
````

### 🤖 AI/ML Application: Model Configuration

````python
# Storing model hyperparameters
model_config = {
    "model_name": "ResNet50",
    "learning_rate": 0.001,
    "batch_size": 32,
    "epochs": 100,
    "optimizer": "Adam",
    "dropout": 0.2,
    "input_shape": (224, 224, 3),
    "num_classes": 10
}

print("Model Configuration:")
for key, value in model_config.items():
    print(f"  {key}: {value}")
````

## 📖 2. Accessing Values

````python
print("--- 2. Accessing Values ---")

player = {"name": "Hero", "level": 10, "gold": 500}

# Access by key
print(f"Name: {player['name']}")
print(f"Level: {player['level']}")

# get() method - safer (returns None if key doesn't exist)
health = player.get("health")  # Returns None
print(f"Health: {health}")

# get() with default value
health = player.get("health", 100)  # Returns 100 if not found
print(f"Health (with default): {health}")

# Using [] on missing key raises KeyError
# player["mana"]  # Would raise KeyError!
````

---
## 📖 3. Modifying Dictionaries

````python
print("--- 3. Modifying Dictionaries ---")

player = {"name": "Hero", "level": 10}
print("Original:", player)

# Add new key-value pair
player["gold"] = 500
print("After adding gold:", player)

# Update existing value
player["level"] = 11
print("After leveling up:", player)

# update() - Add/update multiple at once
player.update({"health": 100, "mana": 50, "level": 12})
print("After update():", player)

# Remove with del
del player["mana"]
print("After del mana:", player)

# pop() - Remove and return value
gold = player.pop("gold")
print(f"Popped gold: {gold}, Player now: {player}")
````

### 🤖 AI/ML Application: Dynamic Config Updates

````python
# Updating training config during experiments
config = {
    "learning_rate": 0.001,
    "batch_size": 32
}
print("Initial config:", config)

# Experiment 1: Try larger batch
config["batch_size"] = 64
config["experiment"] = "exp_001"
print("Experiment 1:", config)

# Experiment 2: Learning rate schedule
config.update({
    "learning_rate": 0.0001,
    "lr_scheduler": "cosine",
    "experiment": "exp_002"
})
print("Experiment 2:", config)
````

## 📖 4. Dictionary Methods

````python
print("--- 4. Dictionary Methods ---")

stats = {"hp": 100, "mp": 50, "str": 25, "dex": 20}

# keys() - Get all keys
print(f"Keys: {list(stats.keys())}")

# values() - Get all values
print(f"Values: {list(stats.values())}")

# items() - Get key-value pairs as tuples
print(f"Items: {list(stats.items())}")

# Check if key exists
print(f"'hp' in stats? {'hp' in stats}")
print(f"'int' in stats? {'int' in stats}")

# Length
print(f"Number of stats: {len(stats)}")

# clear() - Remove all items
copy = stats.copy()
copy.clear()
print(f"After clear: {copy}")
````

## 📖 5. Iterating Through Dictionaries

````python
print("--- 5. Iterating ---")

inventory = {"sword": 1, "potion": 5, "gold": 100, "key": 3}

# Iterate over keys (default)
print("Keys:")
for item in inventory:
    print(f"  {item}")

# Iterate over values
print("\nValues:")
for quantity in inventory.values():
    print(f"  {quantity}")

# Iterate over key-value pairs (most common)
print("\nInventory:")
for item, qty in inventory.items():
    print(f"  {item}: {qty}")
````

### 🤖 AI/ML Application: Training Metrics

````python
# Storing and displaying training metrics
metrics = {
    "accuracy": 0.92,
    "precision": 0.89,
    "recall": 0.91,
    "f1_score": 0.90,
    "loss": 0.25
}

print("Training Metrics:")
print("-" * 25)
for metric, value in metrics.items():
    if metric == "loss":
        print(f"{metric:12}: {value:.4f}")
    else:
        print(f"{metric:12}: {value*100:.2f}%")

# Calculate average (excluding loss)
score_metrics = [v for k, v in metrics.items() if k != "loss"]
avg_score = sum(score_metrics) / len(score_metrics)
print(f"{'Average':12}: {avg_score*100:.2f}%")
````

## 📖 6. Nested Dictionaries

````python
print("--- 6. Nested Dictionaries ---")

# Game with multiple players
game_state = {
    "player1": {
        "name": "Hero",
        "level": 25,
        "inventory": ["sword", "shield"]
    },
    "player2": {
        "name": "Mage",
        "level": 22,
        "inventory": ["staff", "robe"]
    }
}

# Access nested values
p1_name = game_state["player1"]["name"]
p1_level = game_state["player1"]["level"]
print(f"Player 1: {p1_name}, Level {p1_level}")

# Modify nested value
game_state["player1"]["level"] += 1
print(f"Player 1 leveled up to {game_state['player1']['level']}!")
````

### 🤖 AI/ML Application: Model Architecture Config

````python
# Complex model configuration
model_arch = {
    "encoder": {
        "type": "ResNet",
        "layers": 50,
        "pretrained": True
    },
    "decoder": {
        "type": "UNet",
        "channels": [256, 128, 64]
    },
    "training": {
        "optimizer": "Adam",
        "learning_rate": 0.001,
        "loss": "CrossEntropy"
    }
}

# Access nested config
lr = model_arch["training"]["learning_rate"]
encoder_type = model_arch["encoder"]["type"]
print(f"Using {encoder_type} encoder with LR={lr}")

# Safely access with get()
dropout = model_arch["training"].get("dropout", 0.0)
print(f"Dropout: {dropout}")
````

# 📝 Key Takeaways

| Operation | Syntax | Example |
|-----------|--------|--------|
| Create | `{}` | `{"key": value}` |
| Access | `d[key]` | `player["name"]` |
| Safe Access | `.get(key)` | `d.get("x", default)` |
| Add/Update | `d[key] = val` | `d["new"] = 5` |
| Delete | `del d[key]` | `del d["old"]` |
| Keys | `.keys()` | `d.keys()` |
| Values | `.values()` | `d.values()` |
| Items | `.items()` | `d.items()` |

**ML Connections:**
- Store hyperparameters: `{"lr": 0.001, "batch": 32}`
- Training metrics: `{"accuracy": 0.92, "loss": 0.25}`
- Model configs: Nested dictionaries
- Label counting: `{label: count}`
