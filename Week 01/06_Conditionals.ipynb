## 📖 What are Conditionals?
**Conditionals** let your code make decisions. Think of them as the "Choose Your Own Adventure" mechanics in games!

**Key Concepts:**
- `if` - Do something IF condition is True
- `elif` - Else if (check another condition)
- `else` - If nothing else matches
- Indentation matters in Python!

---
## 📖 1. Basic If Statement

````python
print("--- 1. Basic If Statement ---")

player_hp = 75

# Simple if - runs only if condition is True
if player_hp > 50:
    print("Player is healthy!")

# If condition is False, nothing happens
if player_hp > 100:
    print("This won't print because 75 is not > 100")

print("Game continues...")
````

---
## 📖 2. If-Else Statement

````python
print("--- 2. If-Else ---")

player_hp = 30

if player_hp > 50:
    print("✓ Player is healthy!")
else:
    print("⚠️ Player needs healing!")

# Another example
has_key = False

if has_key:
    print("Door opens...")
else:
    print("The door is locked!")
````

---
## 📖 3. If-Elif-Else Chain

````python
print("--- 3. If-Elif-Else Chain ---")

player_hp = 25

if player_hp >= 80:
    status = "Healthy 💚"
elif player_hp >= 50:
    status = "Injured 💛"
elif player_hp >= 20:
    status = "Critical 🧡"
else:
    status = "Near Death ❤️"

print(f"HP: {player_hp} - Status: {status}")

# Try changing player_hp and run again!
````

### 🤖 AI/ML Application: Classification Thresholds

````python
# Binary classification with probability threshold
probability = 0.73
threshold = 0.5

if probability >= threshold:
    prediction = "Positive"
else:
    prediction = "Negative"

print(f"Probability: {probability}, Threshold: {threshold}")
print(f"Prediction: {prediction}")

# Multi-class confidence classification
confidence = 0.82

if confidence >= 0.9:
    certainty = "Very High"
elif confidence >= 0.7:
    certainty = "High"
elif confidence >= 0.5:
    certainty = "Medium"
else:
    certainty = "Low - needs review"

print(f"\nConfidence: {confidence} → {certainty}")
````

## 📖 4. Comparison Operators Review

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `x == 5` |
| `!=` | Not equal | `x != 5` |
| `>` | Greater than | `x > 5` |
| `<` | Less than | `x < 5` |
| `>=` | Greater or equal | `x >= 5` |
| `<=` | Less or equal | `x <= 5` |

````python
print("--- 4. Comparison Operators ---")

score = 85

print(f"score == 85: {score == 85}")
print(f"score != 100: {score != 100}")
print(f"score > 80: {score > 80}")
print(f"score < 90: {score < 90}")
print(f"score >= 85: {score >= 85}")
print(f"score <= 84: {score <= 84}")
````

## 📖 5. Logical Operators (and, or, not)

````python
print("--- 5. Logical Operators ---")

level = 10
has_key = True
is_boss_defeated = False

# AND - Both must be True
can_enter_dungeon = level >= 5 and has_key
print(f"Level {level}, Has Key: {has_key}")
print(f"Can enter dungeon (AND): {can_enter_dungeon}")

# OR - At least one must be True
can_proceed = has_key or is_boss_defeated
print(f"\nCan proceed (OR): {can_proceed}")

# NOT - Flips the value
door_locked = not has_key
print(f"Door locked (NOT has_key): {door_locked}")
````

### 🤖 AI/ML Application: Data Validation

````python
# Validate data before training
num_samples = 1000
has_labels = True
no_missing = True
is_normalized = False

# Check minimum requirements
min_samples = 100

if num_samples >= min_samples and has_labels and no_missing:
    print("✓ Data is ready for training!")
    
    if not is_normalized:
        print("⚠️ Warning: Data is not normalized")
else:
    print("✗ Data validation failed!")
    
    if num_samples < min_samples:
        print(f"  - Need at least {min_samples} samples")
    if not has_labels:
        print("  - Missing labels")
    if not no_missing:
        print("  - Contains missing values")
````

## 📖 6. Nested Conditionals

````python
print("--- 6. Nested Conditionals ---")

player_class = "Mage"
level = 15
has_staff = True

if player_class == "Mage":
    print("You are a Mage!")
    
    if level >= 10:
        print("  You can learn advanced spells!")
        
        if has_staff:
            print("  ✨ Fire spell unlocked!")
        else:
            print("  You need a staff to cast spells")
    else:
        print("  Keep training to unlock spells")
else:
    print("You are not a Mage")
````

## 📖 7. Ternary Operator (One-Line If)

````python
print("--- 7. Ternary Operator ---")

# Syntax: value_if_true if condition else value_if_false

hp = 75
status = "Healthy" if hp > 50 else "Injured"
print(f"HP: {hp}, Status: {status}")

# Great for quick assignments
score = 85
passed = "Pass" if score >= 60 else "Fail"
print(f"Score: {score}, Result: {passed}")

# ML example: Quick classification
probability = 0.72
label = 1 if probability >= 0.5 else 0
print(f"Probability: {probability}, Label: {label}")
````

### 🤖 AI/ML Application: Feature Engineering

````python
# Creating binary features from continuous values
age = 25
income = 55000
credit_score = 720

# Create binary features
is_young = 1 if age < 30 else 0
high_income = 1 if income > 50000 else 0
good_credit = 1 if credit_score >= 700 else 0

print("Original values:")
print(f"  Age: {age}, Income: {income}, Credit: {credit_score}")
print("\nBinary features:")
print(f"  is_young: {is_young}")
print(f"  high_income: {high_income}")
print(f"  good_credit: {good_credit}")

# Calculate simple risk score
risk_score = is_young + high_income + good_credit
risk_level = "Low" if risk_score >= 2 else "High"
print(f"\nRisk Score: {risk_score}/3 → {risk_level} Risk")
````

## 📖 8. Membership Testing (in, not in)

````python
print("--- 8. Membership Testing ---")

inventory = ["sword", "shield", "potion"]

if "sword" in inventory:
    print("You have a sword!")

if "bow" not in inventory:
    print("You don't have a bow")

# Works with strings too
message = "Welcome to the game"
if "game" in message:
    print("This is a game message")
````
