## 📖 What is a Function?
A **function** is a reusable block of code that performs a specific task. Think of it as crafting a skill that you can use anytime!

**Benefits:**
- Code reuse (DRY - Don't Repeat Yourself)
- Better organization
- Easier testing and debugging
- Cleaner, more readable code

## 📖 1. Defining a Function

````python
print("--- 1. Defining Functions ---")

# Define with 'def' keyword
def greet():
    print("Hello, Player!")

# Call the function
greet()
greet()  # Can call multiple times!
````

## 📖 2. Parameters (Inputs)

````python
print("--- 2. Parameters ---")

# Function with parameter
def attack(enemy_name):
    print(f"You attacked {enemy_name}!")

attack("Goblin")
attack("Dragon")

# Multiple parameters
def attack_with_weapon(enemy, weapon):
    print(f"You attacked {enemy} with {weapon}!")

attack_with_weapon("Orc", "Sword")
attack_with_weapon("Troll", "Hammer")
````

## 📖 3. Return Values (Outputs)

````python
print("--- 3. Return Values ---")

# Return a value
def calculate_damage(base, multiplier):
    total = base * multiplier
    return total

damage = calculate_damage(50, 1.5)
print(f"Damage dealt: {damage}")

# Use directly in expressions
print(f"Critical hit: {calculate_damage(50, 2.0)}")
````

### 🤖 AI/ML Application: Preprocessing Functions

````python
# Reusable normalization function
def normalize(values):
    """Normalize values to 0-1 range"""
    min_val = min(values)
    max_val = max(values)
    return [(v - min_val) / (max_val - min_val) for v in values]

data1 = [10, 20, 30, 40, 50]
data2 = [100, 200, 300]

print(f"Original: {data1}")
print(f"Normalized: {normalize(data1)}")

print(f"\nOriginal: {data2}")
print(f"Normalized: {normalize(data2)}")
````

## 📖 4. Default Parameters

````python
print("--- 4. Default Parameters ---")

def heal(player, amount=25):
    print(f"Healed {player} for {amount} HP!")

# Use default
heal("Hero")

# Override default
heal("Hero", 50)

# Named parameters
heal(amount=100, player="Mage")
````

### 🤖 AI/ML Application: Flexible Model Config

````python
def train_model(data, epochs=10, learning_rate=0.001, batch_size=32):
    """Simulate model training with configurable hyperparameters"""
    print(f"Training on {len(data)} samples")
    print(f"  Epochs: {epochs}")
    print(f"  Learning Rate: {learning_rate}")
    print(f"  Batch Size: {batch_size}")
    return 0.95  # Simulated accuracy

# Use defaults
data = list(range(1000))
accuracy = train_model(data)
print(f"Accuracy: {accuracy}")

# Override some parameters
print()
accuracy = train_model(data, epochs=50, learning_rate=0.0001)
print(f"Accuracy: {accuracy}")
````

## 📖 5. Returning Multiple Values

````python
print("--- 5. Multiple Return Values ---")

def loot_chest():
    gold = 100
    item = "Magic Ring"
    return gold, item  # Returns a tuple

# Unpack the results
gold, item = loot_chest()
print(f"You got {gold} gold and a {item}!")
````

### 🤖 AI/ML Application: Evaluation Metrics

````python
def evaluate_model(actual, predicted):
    """Calculate multiple metrics at once"""
    correct = sum(1 for a, p in zip(actual, predicted) if a == p)
    total = len(actual)
    
    accuracy = correct / total
    error_rate = 1 - accuracy
    
    return accuracy, error_rate, correct, total

actual = [1, 0, 1, 1, 0, 1, 0]
predicted = [1, 0, 0, 1, 0, 1, 1]

acc, err, correct, total = evaluate_model(actual, predicted)

print(f"Results:")
print(f"  Correct: {correct}/{total}")
print(f"  Accuracy: {acc:.2%}")
print(f"  Error Rate: {err:.2%}")
````

## 📖 6. Docstrings (Documentation)

````python
print("--- 6. Docstrings ---")

def calculate_accuracy(correct, total):
    """
    Calculate accuracy percentage.
    
    Args:
        correct: Number of correct predictions
        total: Total number of predictions
        
    Returns:
        float: Accuracy as a decimal (0-1)
    """
    return correct / total

# Use the function
acc = calculate_accuracy(85, 100)
print(f"Accuracy: {acc:.2%}")

# View documentation
print(f"\nDocstring: {calculate_accuracy.__doc__}")
````

## 📖 7. Scope (Local vs Global)

````python
print("--- 7. Scope ---")

# Global variable
player_level = 10

def show_level():
    # Can READ global variables
    print(f"Player is level {player_level}")

def level_up():
    # To MODIFY global, use 'global' keyword
    global player_level
    player_level += 1
    print(f"Leveled up to {player_level}!")

show_level()
level_up()
show_level()
````

    min_val = min(numbers)
    max_val = max(numbers)

min_val, max_val, avg = get_stats([10, 20, 30, 40, 50])

# 📝 Key Takeaways

| Concept | Syntax |
|---------|--------|
| Define | `def name():` |
| Parameters | `def name(a, b):` |
| Defaults | `def name(a=1):` |
| Return | `return value` |
| Multiple returns | `return a, b` |
| Docstring | `"""Description"""` |

**ML Connections:**
- Preprocessing: `normalize()`, `standardize()`
