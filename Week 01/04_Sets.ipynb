## 📖 What is a Set?
A **Set** is an unordered collection of **unique** elements. Think of it as a bag of power-ups where duplicates are automatically removed!

**Key Properties:**
- Unordered (no indexing)
- **No duplicates** (automatic)
- Mutable (can add/remove)
- Uses curly braces `{}`
- Very fast membership testing

---
## 📖 1. Creating Sets

````python
print("--- 1. Creating Sets ---")

# Create with curly braces
powerups = {"shield", "sword", "potion"}
print("Power-ups:", powerups)

# Duplicates are automatically removed!
items = {"coin", "coin", "gem", "coin", "gem"}
print("Unique items:", items)  # Only 2 items!

# Create from list (removes duplicates)
scores = [85, 90, 85, 95, 90, 100]
unique_scores = set(scores)
print(f"Scores: {scores}")
print(f"Unique scores: {unique_scores}")

# Empty set (use set(), not {}!)
empty_set = set()  # {} creates an empty dict!
print(f"Empty set: {empty_set}, Type: {type(empty_set).__name__}")
````

### 🤖 AI/ML Application: Unique Labels & Vocabulary

````python
# Finding unique class labels in a dataset
all_labels = ["cat", "dog", "cat", "bird", "dog", "cat", "fish", "dog"]
unique_labels = set(all_labels)
num_classes = len(unique_labels)

print(f"All labels ({len(all_labels)}): {all_labels}")
print(f"Unique labels ({num_classes}): {unique_labels}")

# Building vocabulary from text
sentence = "the cat sat on the mat"
words = sentence.split()
vocabulary = set(words)

print(f"\nWords: {words}")
print(f"Vocabulary: {vocabulary}")
print(f"Vocabulary size: {len(vocabulary)}")
````

## 📖 2. Adding & Removing Elements

````python
print("--- 2. Adding & Removing ---")

skills = {"fire", "ice"}
print("Starting skills:", skills)

# add() - Add single element
skills.add("lightning")
print("After add('lightning'):", skills)

# Adding duplicate does nothing
skills.add("fire")
print("After add('fire') again:", skills)

# update() - Add multiple elements
skills.update(["earth", "water"])
print("After update():", skills)

# remove() - Raises error if not found
skills.remove("ice")
print("After remove('ice'):", skills)

# discard() - No error if not found
skills.discard("ice")  # No error even though 'ice' was removed
print("After discard('ice'):", skills)

# pop() - Remove and return random element
popped = skills.pop()
print(f"Popped: {popped}, Remaining: {skills}")
````

## 📖 3. Set Operations

Sets support mathematical operations:

| Operation | Symbol | Method | Description |
|-----------|--------|--------|-------------|
| Union | `\|` | `.union()` | All elements from both |
| Intersection | `&` | `.intersection()` | Elements in both |
| Difference | `-` | `.difference()` | In first, not in second |
| Symmetric Diff | `^` | `.symmetric_difference()` | In one, not both |

````python
print("--- 3. Set Operations ---")

team_a = {"Ali", "Hamza", "Zain"}
team_b = {"Zain", "Umer", "Bilal"}

print(f"Team A: {team_a}")
print(f"Team B: {team_b}")

# Union - Everyone
all_players = team_a | team_b
print(f"\nAll players (union): {all_players}")

# Intersection - In both teams
both_teams = team_a & team_b
print(f"In both teams (intersection): {both_teams}")

# Difference - Only in Team A
only_a = team_a - team_b
print(f"Only in Team A (difference): {only_a}")

# Symmetric Difference - In one but not both
exclusive = team_a ^ team_b
print(f"Exclusive to one team: {exclusive}")
````

### 🤖 AI/ML Application: Feature Comparison

````python
# Comparing features between two models
model_v1_features = {"age", "income", "location", "gender"}
model_v2_features = {"age", "income", "education", "job_title"}

print("Model v1 features:", model_v1_features)
print("Model v2 features:", model_v2_features)

# Common features
common = model_v1_features & model_v2_features
print(f"\nCommon features: {common}")

# Features removed in v2
removed = model_v1_features - model_v2_features
print(f"Removed in v2: {removed}")

# New features in v2
added = model_v2_features - model_v1_features
print(f"Added in v2: {added}")

# All features ever used
all_features = model_v1_features | model_v2_features
print(f"All features: {all_features}")
````

## 📖 4. Membership Testing (Fast!)

````python
print("--- 4. Membership Testing ---")

valid_moves = {"up", "down", "left", "right"}

# Check membership with 'in'
move = "up"
if move in valid_moves:
    print(f"'{move}' is a valid move!")

move2 = "diagonal"
if move2 not in valid_moves:
    print(f"'{move2}' is NOT a valid move!")

# Sets are MUCH faster than lists for membership testing
# For large datasets, this matters A LOT!
````

### 🤖 AI/ML Application: Stopword Filtering

````python
# Fast stopword removal using sets
stopwords = {"the", "a", "an", "is", "are", "was", "on", "in", "at"}

sentence = "the cat is on the mat"
words = sentence.split()

# Filter out stopwords
filtered = [word for word in words if word not in stopwords]

print(f"Original: {words}")
print(f"Stopwords: {stopwords}")
print(f"Filtered: {filtered}")

# This is O(1) per word because sets use hashing!
````

## 📖 5. Set Comparisons

````python
print("--- 5. Set Comparisons ---")

basic_skills = {"attack", "defend"}
advanced_skills = {"attack", "defend", "magic", "stealth"}

# issubset() - Is A contained in B?
print(f"basic ⊆ advanced? {basic_skills.issubset(advanced_skills)}")

# issuperset() - Does A contain B?
print(f"advanced ⊇ basic? {advanced_skills.issuperset(basic_skills)}")

# isdisjoint() - No common elements?
set1 = {1, 2, 3}
set2 = {4, 5, 6}
set3 = {3, 4, 5}
print(f"set1 ∩ set2 = ∅? {set1.isdisjoint(set2)}")
print(f"set1 ∩ set3 = ∅? {set1.isdisjoint(set3)}")
````

# 📝 Key Takeaways

| Operation | Syntax | Example |
|-----------|--------|--------|
| Create | `{}` or `set()` | `{1, 2, 3}` |
| Add | `.add(x)` | `s.add(4)` |
| Remove | `.remove(x)` | `s.remove(2)` |
| Union | `\|` | `a \| b` |
| Intersection | `&` | `a & b` |
| Difference | `-` | `a - b` |
| Check member | `in` | `x in s` |

**ML Connections:**
- Find unique class labels: `set(labels)`
- Build vocabulary: `set(words)`
- Fast stopword filtering
- Compare model features
